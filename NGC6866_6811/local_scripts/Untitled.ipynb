```python
import numpy as np
from astropy.table import QTable, vstack

# Read both catalogs

cat_6811 = QTable.read("rcat_ngc6811_v0.fits")
cat_6866 = QTable.read("rcat_ngc6866_v0.fits")

# Combine them into one table

catalog = vstack([cat_6811, cat_6866])

# Total number of rows

n_rows = len(catalog)

# Number of unique Gaia IDs

unique_ids = np.unique(catalog["GAIAEDR3_ID"])
n_unique = len(unique_ids)

print(f"Total rows in catalog: {n_rows}")
print(f"Unique Gaia IDs: {n_unique}")

# Compare them

if n_rows == n_unique:
    print("\nTable is ONE ROW PER STAR")
    print("You cannot count RV epochs from this table.")
else:
    print("\nTable contains MULTIPLE ROWS PER STAR")
    print("You CAN count RV measurements from this table.")

    # Show an example of repeated stars

    print("\nStars with repeated rows:\n")

    for star_id in unique_ids:

        count = np.count_nonzero(
            catalog["GAIADER3_ID"] == star_id
        )

        if count > 1:
            print(f"{star_id}: {count} rows")
```
