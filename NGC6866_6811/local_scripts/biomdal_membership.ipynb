```python
from astropy.table import Table

bimodal = Table.read("bimodalcheck_200M_phase.csv", format="csv")

bimodal_1 = bimodal[bimodal["bimodal"] == 1]
bimodal_neg1 = bimodal[bimodal["bimodal"] == -1]
bimodal_neg2 = bimodal[bimodal["bimodal"] == -2]

print(f"Stars with bimodal = 1 ({len(bimodal_1)}):")
for row in bimodal_1:
    print(row["id"])

print(f"\nStars with bimodal = -1 ({len(bimodal_neg1)}):")
for row in bimodal_neg1:
    print(row["id"])

print(f"\nStars with bimodal = -2 ({len(bimodal_neg2)}):")
for row in bimodal_neg2:
    print(row["id"])

print("----------------------------")

unimodal1=bimodal[bimodal["unimodal"] == 1]
print(f"Stars with unimodal = 1 ({len(unimodal1)}):")


print(bimodal.colnames, flush=True)
```
