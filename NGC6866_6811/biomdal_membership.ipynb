```python
from astropy.table import Table, vstack

bimodal = Table.read("bimodalcheck_200M_phase.csv", format="csv")
ngc6866 = Table.read("rcat_ngc6866_v0.fits")
ngc6811 = Table.read("rcat_ngc6811_v0.fits")

catalog = vstack([ngc6866, ngc6811], metadata_conflicts="silent")

bimodal_stars = bimodal[bimodal["bimodal"] == 1]

print(f"Found {len(bimodal_stars)} stars with bimodal = 1\n")

for row in bimodal_stars:
    star_id = row["id"]
    matches = catalog[catalog["GAIAEDR3_ID"] == star_id]

    if len(matches) == 0:
        print(f"{star_id}: not found in catalog")
        continue

    m = matches[0]

    mem = bool(m["MemBool"])

    if mem:
        print(f"{star_id}: member")
    else:
        print(f"{star_id}: non-member, HDBscan_Cluster={m['HDBscan_Cluster']}")

print()

unsure = bimodal[bimodal["bimodal"] == -1]

print(f"Found {len(unsure)} stars with bimodal = -1\n")

for row in unsure:
    star_id = row["id"]
    matches = catalog[catalog["GAIAEDR3_ID"] == star_id]

    if len(matches) == 0:
        print(f"{star_id}: not found in catalog")
        continue

    m = matches[0]
    mem = bool(m["MemBool"])

    if mem:
        print(f"{star_id}: member")
    else:
        print(f"{star_id}: non-member, HDBscan_Cluster={m['HDBscan_Cluster']}")

print()

mcmc_used = unsure["MCMC"] == 1
unsure_mcmc = unsure[mcmc_used]

print(f"Found {len(unsure_mcmc)} stars with bimodal = -1 AND MCMC used\n")
```
