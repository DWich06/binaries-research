```python
import numpy as np
import thejoker as tj

star_id = 2076065836598553728

sample_file = f"/home/dwich2006/binaries-research/NGC6866_6811/data/{star_id}/rejection_samples_MCMC_200.0M_{star_id}_new.hdf5"

samples = tj.JokerSamples.read(sample_file)

P = samples["P"].to_value("day")
e = samples["e"]

print("Number of samples:", len(samples))
```

```python
print("P min:", np.min(P))
print("P max:", np.max(P))
print("P stdev:", np.std(P))
print("log10(P) min:", np.min(np.log10(P)))
print("log10(P) max:", np.max(np.log10(P)))
print("log10(P) stdev:", np.std(np.log10(P)))

print()

print("e min:", np.min(e))
print("e max:", np.max(e))
print("e stdev:", np.std(e))
```

```python
import matplotlib.pyplot as plt

plt.hist(np.log10(P), bins=20)
plt.xlabel("log10(P/day)")
plt.ylabel("Count")
plt.show()

print("The weird number means that the actual values are the display tick * 1e-6 - 9.8607e-1")
```
