```python
import numpy as np
from astropy.table import Table, join
import matplotlib.pyplot as plt
```

```python
def unimodal_hdbscan(unimodal_file, fits_file):
    uni = Table.read(unimodal_file)
    cluster = Table.read(fits_file)

    if "id" in uni.colnames and "GAIAEDR3_ID" not in uni.colnames:
        uni = uni.copy()
        uni.rename_column("id", "GAIAEDR3_ID")

    merged = join(uni, cluster, keys="GAIAEDR3_ID", join_type="inner")

    good = (
        np.isfinite(merged["dist"]) &
        np.isfinite(merged["HDBscan_Cluster"]) &
        (merged["dist"] > 0)
    )
    merged = merged[good]

    # Condition
    mask = (merged["unimodal"] == 1) & (merged["HDBscan_Cluster"] == -2)
    subset = merged[mask]

    print("\nIDs with unimodal == 1 and HDBscan_Cluster == -2:\n")
    if len(subset) == 0:
        print("0")
    else:
        for star_id in subset["GAIAEDR3_ID"]:
            print(star_id)
```

```python
unimodal_hdbscan("unimodalcheck_200M_new.csv", "rcat_ngc6866_v0.fits")
```
